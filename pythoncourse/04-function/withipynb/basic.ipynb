```python
def agentState(state):
    if state == "error":
        return "agent: Something went wrong"
    elif state == "task":
        return "agent: Executing task"
    elif state == "needs_tool":
        return "agent: Needs a tool to complete the task"
    elif state == "tool_name":
        return "agent: Tool needed is " + state["tool_name"]
    else:
        return "agent: State not recognized"
```

```python
print(agentState("needs_tool"))
```
