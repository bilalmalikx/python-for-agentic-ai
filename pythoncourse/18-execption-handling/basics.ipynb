```python
class AgeTooSmallError(Exception):
    """Custom exception for age less than required"""
    def __init__(self, message="Age 18 se kam nahi honi chahiye"):
        self.message = message
        super().__init__(self.message)
```

```python
def check_age(age):
    if age < 18:
        raise AgeTooSmallError()  # Custom exception raise kar rahe hain
    else:
        print("Access allowed.")

try:
    check_age(16)
except AgeTooSmallError as e:
    print("Exception:", e)
```

```python
# Custom Exception class
class Experiance(Exception):
    pass

# Function to check experience
def check_exp(exp):
    if exp > 3:
        raise Experiance("Experience is too high!")  # raise with a message
    else:
        print("Access Allowed")

# Try-Except block
try:
    check_exp(4)
except Experiance as x:
    print("Exp:", x)
```
