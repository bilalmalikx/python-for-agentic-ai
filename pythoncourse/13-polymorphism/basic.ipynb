```python
# Method overriding
# Method overrinding alolow the clild class to provide a specific implementation of a method that
# is already defined in its parent class. This allows for dynamic behavior based on the object type at runtime.

# Base class
# Method overriding

class Animal:
    def speak(self):
        return "Animal speaks"

class dog(Animal):
    def speak(self):
        return "Dog barks"

class cat(Animal):
    def speak(self):
        return "Cat meows"

def animal_speak(animal):  # lowercase 'a'
    print(animal.speak())

Dog = dog()
Cat = cat()

print(Dog.speak())  # Output: Dog barks
print(Cat.speak())  # Output: Cat meows

animal_speak(Dog)  
```

```python
# base class
class Shape:
    def area(self):
        return "Area of the shape"
    
# Derived Class
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
# derived class
class circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * self.radius * self.radius
    
def print_area(shape):
    print(f"the area is {shape.area()}")
    
    
rectangle = Rectangle(5, 10)
circle1 = circle(7)

print_area(rectangle)  # Output: 50
print_area(circle) # Output: 153.86
```

```python
# Pollymorphisan with abstract class
from abc import ABC, abstractmethod

# Define an abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    
class Car(Vehicle):
    def start_engine(self):
        return "car engine started"
    
class Bike(Vehicle):
    def start_engine(self):
        return "bike engine started"
    
def start_vehicle(vehicle):
    print(vehicle.start_engine())
    
car1 = Car()
bike1 = Bike()

start_vehicle(car1)
start_vehicle(bike1)
```

```python
class Employee:
    def __init__(self, name, salary, base_salary=50000):
        self.name = name
        self.salary = salary
        self.base_salary = base_salary
    def calculate_salary(self):
        return self.salary + self.base_salary        

class Manager(Employee):
    def __init__(self, name, salary, base_salary=50000, extra_allowance=50):
        super().__init__(name, salary, base_salary)
        self.extra_allowance = extra_allowance
    
    def calculate_salary(self):
        return super().calculate_salary() + self.extra_allowance
    
class Developer(Employee):
    def __init__(self, name, salary, base_salary=30000, bonus=30):
        super().__init__(name, salary, base_salary)
        self.bonus = bonus
    def calculate_salary(self):
        return super().calculate_salary() + self.bonus
    
class Intern(Employee):
    def __init__(self, name, salary, base_salary=20000, bonus=20):
        super().__init__(name, salary, base_salary)
        self.bonus = bonus
    def calculate_salary(self):
        return super().calculate_salary() + self.bonus
    
manager = Manager("Alice", 10000)
developer = Developer("Bob", 8000)
intern = Intern("Charlie", 5000)
print(f"Manager {manager.name}'s salary: {manager.calculate_salary()}")
print(f"Developer {developer.name}'s salary: {developer.calculate_salary()}")    
```
