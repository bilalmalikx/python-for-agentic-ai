```python
def is_not_patakha(item):
    return item != 'patakha'

tokri = ['gajar', 'aalu', 'patakha', 'mooli', 'patakha', 'pyaaz']

result = filter(is_not_patakha, tokri)

print(list(result))  # Output: ['gajar', 'aalu', 'mooli', 'pyaaz']
```

```python
name = ["bilal","malik","midrar","malik"]

def only_brothers_name(item):
    return item != "malik"


result = filter(only_brothers_name,name)
print(list(result))
```
