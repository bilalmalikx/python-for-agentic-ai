```python
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([6, 7, 8, 9, 10])
print("addtion = ", arr1 + arr2) # add two arrays
print("subtraction = ", arr1 - arr2) # subtract two arrays
print("multiplication = ", arr1 * arr2) # multiply two arrays
print("division = ", arr1 / arr2) # divide two arrays
```

```python
# universal function÷//

arr1 = np.array([1, 2, 3, 4, 5])
# squre root
print("spure root = ", np.sqrt(arr1)) # square root of each element in the array

print(np.exp(arr1)) # exponential of each element in the array

print(np.sin(arr1))

print(np.log(arr1))
```
