```python
names: list[str] = ["bilal","midrar","sara","makhi","ali","hassan","mohsin","ahmed","zainab","fatima","khadija","abdullah","omar","yusuf","nour","layla","mariam","sami","huda","khalid","laila"]

for name in names:
    if "makhi" in name:
        print(f"{name} in the list")
    elif "bilal" in name:
        print(f"{name} is in the list")
    else:
        print(f"{name} is not in the list")
```

```python
found = False
for name in names:
    if "makhi" in name:
        found = True
        break 

if found:
    print("makhi is in the list")
else:
    print("makhi is not in the list")
```
