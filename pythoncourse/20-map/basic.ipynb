```python
def multiply_by_two(num):
    return num * 2

numbers = [1, 2, 3, 4, 5]

result = map(multiply_by_two, numbers)

print(list(result))  # Output: [2, 4, 6, 8, 10]
```

```python
numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * 2, numbers)

print(list(result))  # Output: [2, 4, 6, 8, 10]
```

```python
words = ['bilal', 'malik', 'python']

result = map(str.upper, words)

print(list(result))  # Output: ['BILAL', 'MALIK', 'PYTHON']
```

```python
def multiplay_by_two(num):
    return num 

numbers = [2, 4, 6, 8, 10]
result = map(multiplay_by_two, numbers)

print(list(result))  # Output: [4, 8, 12, 16, 20]
```
