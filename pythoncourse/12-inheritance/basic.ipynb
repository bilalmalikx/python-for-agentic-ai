# 👇 Difference between Parent, Child, and Other Classes

| Term         | Description                                                               | Example                                     |
|--------------|---------------------------------------------------------------------------|---------------------------------------------|
| Parent Class | Ek base class jo common functionality define karti hai                    | class Animal: def sound(self): pass         |
| Child Class  | Parent class se inherit karti hai aur apne methods ya features add karti hai | class Dog(Animal): def bark(self): pass     |
| Other Class  | Koi aur class jo alag logic rakhti ho, inherit na kare                    | class Car: def drive(self): pass            |


```python
class car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
        
    def drive(self):
        print(f"the person will drive the {self.enginetype} car")
        
```

```python
car1 = car(4, 4, "petrol")
car1.drive()
```

```python
class tesla(car):
    def __init__(self, windows, doors, enginetype,is_self_driving):
        super().__init__(windows, doors, enginetype)
        self.is_self_driving = is_self_driving
        
    def self_driving(self):
        print(f"Tesla supports self driving: {self.is_self_driving}")
        
```

```python
tesla1 = tesla(4,4,"electric", True)
tesla1.self_driving()
```

```python
# multi inheritace
# when a class is more the the base class

# Base class 1
class animal:
    def __init__(self,name):
        self.name = name
        
    def speak(self):
        print("subclass must be implemented these kind of methinds")
        
# Base class 2
class pet:
    def __init__(self,owner):
        self.owner = owner
        
# Derived class
class dog(animal, pet):
    def __init__(self, name, owner):
        animal.__init__(self,name)
        pet.__init__(self,owner)
        
    def speak(self):
        print(f"{self.name} says Woof!")
        
dog1 = dog("Buddy", "Alice")
dog1.speak()
print(f"Owner: {dog1.owner}")
```

```python
class FirstName:
    def __init__(self, first_name):
        self.first_name = first_name
        
class LastName:
    def __init__(self, last_name):
        self.last_name = last_name

class FullName(FirstName, LastName):
    def __init__(self, first_name, last_name):
        FirstName.__init__(self, first_name)
        LastName.__init__(self, last_name)

    def greet(self):
        print(f"My full name is {self.first_name} {self.last_name}")

# Object bana ke greet call karte hain
fullname = FullName("Bilal", "Malik")
fullname.greet()
```

```python
class Employee:
    def __init__(self,name, id, department):
        self.name = name
        self.id = id
        self.department = department
        
    def show_details(self):
        return f"my name is {self.name}, and my id is {self.id} and my department is {self.department}"
    
class Manager(Employee):
    def __init__(self, name, id, department, team_size):
        super().__init__(name, id, department)
        self.team_size = team_size
        
    def show_details(self):
        return f"{super().show_details()} and I manage a team of {self.team_size} people"
    
manager = Manager("Bilal", 101, "IT", 5)
print(manager.show_details())
```
