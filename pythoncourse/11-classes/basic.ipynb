| Feature            | Class                                                       | Object                                                     |
|--------------------|-------------------------------------------------------------|-------------------------------------------------------------|
| Definition         | A blueprint or template to create objects                   | An instance of a class                                      |
| Purpose            | To define properties (attributes) and behaviors (methods)   | To represent a real entity with specific values             |
| Memory Allocation  | Does not occupy memory until instantiated                   | Occupies memory when created                                |
| Creation Syntax    | `class Car:`                                                | `my_car = Car()`                                            |
| Reusability        | Can be used to create multiple objects                      | Represents a single, specific instance                      |
| Example            | `class Dog:`                                                | `dog1 = Dog("Tommy")`                                       |
| Real-life Analogy  | Design of a car                                             | A car made from that design                                 |

```python
class time:
    def __init__(self,sec,minute):
        self.sec = sec
        self.minute = minute
        
    def total_spent_time(self):
        return self.sec + (self.minute * 60)
    
    
time1 = time(30,2)
print(time1.total_spent_time())
```

```python
class full_name:
    def __init__(self,first_name,last_name):
        self.first_name = first_name
        self.last_name = last_name
        
    def full_name(self):
        return f'My full name is {self.first_name},{self.last_name}'
    
person = full_name("bilal","malik")
print(person.full_name())
    
```
